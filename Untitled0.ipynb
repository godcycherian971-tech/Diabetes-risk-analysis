```python
import pandas as pd


df = pd.read_csv('diabetes_data_upload[1].csv')


df.head()
```

```python

print("Missing values in data:")
print(df.isnull().sum())

print("\n-------------------------------\n")


print("Class Distribution:")
print(df['class'].value_counts())
```

```python
pd.crosstab(df['visual blurring'], df['class'])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt


plt.figure(figsize=(7,5))
sns.countplot(x='visual blurring', hue='class', data=df, palette='Set2')
plt.title('Visual Blurring vs Diabetes Risk')
plt.xlabel('Visual Blurring (Yes/No)')
plt.ylabel('Patient Count')
plt.show()
```
